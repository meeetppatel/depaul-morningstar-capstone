# Task 1 Best Model Deep Analysis

This notebook analyzes only the selected best Task 1 pipeline:

**v8 six-stream reranker** for 8-digit parent industry classification.

It intentionally excludes failed experiments from the package. The comparison tables may include successful predecessor streams only to show why v8 was selected.

```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix

def find_package_task_dir(expected_file):
    cwd = Path.cwd().resolve()
    candidates = [cwd, cwd / "task1", cwd.parent, cwd.parent.parent, cwd / "final_best_models_package" / "task1"]
    for candidate in candidates:
        if (candidate / "models" / expected_file).exists():
            return candidate
    raise FileNotFoundError(f"Could not find packaged models/{expected_file}. Current working directory: {cwd}")

BASE = find_package_task_dir("task1_v8_test_predictions.csv")
MODEL_DIR = BASE / "models"
OUT = BASE / "analysis_outputs"
OUT.mkdir(exist_ok=True)

plt.style.use("default")
pd.set_option("display.max_columns", 120)
pd.set_option("display.max_colwidth", 160)
print("Using model directory:", MODEL_DIR)
```

```python
pred = pd.read_csv(MODEL_DIR / "task1_v8_test_predictions.csv", dtype=str)
y_true = pred["true"].astype(str).to_numpy()
y_pred = pred["six_stream_reranker_pred"].astype(str).to_numpy()
top10 = pd.Series(y_true).value_counts().head(10).index.astype(str).tolist()
top10_mask = pd.Series(y_true).isin(top10).to_numpy()

metrics = {
    "accuracy": accuracy_score(y_true, y_pred),
    "macro_f1": f1_score(y_true, y_pred, average="macro", zero_division=0),
    "weighted_f1": f1_score(y_true, y_pred, average="weighted", zero_division=0),
    "top10_accuracy": accuracy_score(y_true[top10_mask], y_pred[top10_mask]),
    "top10_macro_f1": f1_score(y_true[top10_mask], y_pred[top10_mask], labels=top10, average="macro", zero_division=0),
}
pd.Series(metrics).to_frame("score")
```

```python
leaderboard_path = MODEL_DIR / "all_task1_prediction_streams_leaderboard.csv"
leaderboard = pd.read_csv(leaderboard_path)
selected = {
    "task1_deberta_mt_with_segdesc_row_probs:argmax_test_leaf": "DeBERTa small MT",
    "task1_v3_deberta_mt_probs_cont:argmax_test_leaf": "DeBERTa base MT",
    "task1_v5_secbert_mt_probs:argmax_test_leaf": "SEC-BERT MT",
    "task1_v8_segfocus_secbert_probs:argmax_test_leaf": "Segment SEC-BERT",
    "task1_v8_test_predictions:six_stream_reranker_pred": "v8 six-stream reranker",
}
comp = leaderboard[leaderboard["model_id"].isin(selected)].copy()
comp["model"] = comp["model_id"].map(selected)
comp = comp.sort_values("macro_f1", ascending=False)
display(comp[["model", "accuracy", "macro_f1", "weighted_f1", "top10_accuracy"]])

fig, ax = plt.subplots(figsize=(9, 4.8))
x = np.arange(len(comp))
w = 0.35
ax.bar(x - w/2, comp["accuracy"], width=w, label="Accuracy", color="#4c78a8")
ax.bar(x + w/2, comp["macro_f1"], width=w, label="Macro F1", color="#f58518")
ax.set_xticks(x)
ax.set_xticklabels(comp["model"], rotation=20, ha="right")
ax.set_ylim(0.60, 0.80)
ax.axhline(0.75, color="#d62728", linestyle="--", linewidth=1.4, label="0.75 overall target")
ax.set_title("Task 1 selected model comparison")
ax.grid(axis="y", alpha=0.25)
ax.legend()
fig.tight_layout()
fig.savefig(OUT / "task1_selected_model_comparison.png", dpi=180)
plt.show()
```

```python
lookup = pd.read_csv(MODEL_DIR / "industry_lookup.csv", dtype=str)
name_map = dict(zip(lookup["industry_code"], lookup["industry_name"]))

report = classification_report(y_true, y_pred, output_dict=True, zero_division=0)
rows = []
for code, vals in report.items():
    if code in {"accuracy", "macro avg", "weighted avg"}:
        continue
    rows.append({
        "industry_code": code,
        "industry_name": name_map.get(code, ""),
        "precision": vals["precision"],
        "recall": vals["recall"],
        "f1": vals["f1-score"],
        "support": vals["support"],
    })
per_class = pd.DataFrame(rows).sort_values(["f1", "support"], ascending=[True, False])
per_class.to_csv(OUT / "task1_v8_per_class_metrics.csv", index=False)

print("Worst classes by F1 with support >= 20")
display(per_class[per_class["support"] >= 20].head(20))
```

```python
worst = per_class[per_class["support"] >= 20].head(15).sort_values("f1")
fig, ax = plt.subplots(figsize=(9, 6))
labels = worst["industry_code"] + " " + worst["industry_name"].fillna("")
ax.barh(labels, worst["f1"], color="#e45756")
ax.set_xlim(0, 1)
ax.axvline(0.75, color="#d62728", linestyle="--", linewidth=1.4, label="0.75 target")
ax.set_title("Task 1 weakest industries by F1, support >= 20")
ax.set_xlabel("F1")
ax.grid(axis="x", alpha=0.25)
ax.legend(loc="lower right")
fig.tight_layout()
fig.savefig(OUT / "task1_weakest_industries_f1.png", dpi=180)
plt.show()
```

## Worst Class Root-Cause Analysis

This section explains why the weakest industries are large misses. It joins the best-model predictions back to the raw Task 1 rows, then measures whether each weak class is mostly failing because of label-boundary confusion, missing/short segment text, non-largest segment context, multi-industry company profiles, or one dominant wrong destination label.


```python
import textwrap

# Root-cause analysis for worst Task 1 classes.
def find_raw_task1_file():
    candidates = [
        Path.cwd() / "task1_gecs_classification_final.csv",
        BASE.parent.parent / "task1_gecs_classification_final.csv",
        BASE.parent / "task1_gecs_classification_final.csv",
        BASE / "task1_gecs_classification_final.csv",
    ]
    for candidate in candidates:
        if candidate.exists():
            return candidate
    raise FileNotFoundError("Could not find task1_gecs_classification_final.csv for row-level root-cause analysis.")

raw_path = find_raw_task1_file()
raw = pd.read_csv(raw_path)
row_idx = pred["test_row_index"].astype(int).to_numpy()
raw_test = raw.iloc[row_idx].copy().reset_index(drop=True)
raw_test["true"] = y_true
raw_test["pred"] = y_pred
raw_test["is_correct"] = raw_test["true"] == raw_test["pred"]

for col in ["CompanyId", "AsOfDate", "SegmentName", "SegmentDescription", "LongProfile"]:
    if col not in raw_test:
        raw_test[col] = ""
    raw_test[col] = raw_test[col].fillna("").astype(str)

raw_test["seg_desc_len"] = raw_test["SegmentDescription"].str.len()
raw_test["profile_len"] = raw_test["LongProfile"].str.len()
raw_test["missing_segment_description"] = raw_test["seg_desc_len"].eq(0)
raw_test["short_segment_description"] = raw_test["seg_desc_len"].between(1, 40)
raw_test["missing_or_short_segment_description"] = raw_test["missing_segment_description"] | raw_test["short_segment_description"]
raw_test["revenue_share_num"] = pd.to_numeric(raw_test.get("revenue_share", np.nan), errors="coerce")
raw_test["largest_share_bool"] = raw_test.get("is_largest_share_segment", "").astype(str).str.lower().isin(["true", "1", "yes"])
raw_test = raw_test.merge(
    raw_test.groupby("CompanyId")["true"].nunique().rename("company_industry_label_count"),
    on="CompanyId",
    how="left",
)

sector_map = dict(zip(lookup["industry_code"], lookup.get("sector_name", "")))
group_map = dict(zip(lookup["industry_code"], lookup.get("industry_group_name", "")))

metric_rows = []
for code, vals in report.items():
    if code in {"accuracy", "macro avg", "weighted avg"}:
        continue
    mask_true = y_true == code
    mask_pred = y_pred == code
    metric_rows.append({
        "industry_code": code,
        "industry_name": name_map.get(code, ""),
        "sector_name": sector_map.get(code, ""),
        "industry_group_name": group_map.get(code, ""),
        "support": int(vals["support"]),
        "correct": int((mask_true & (y_pred == code)).sum()),
        "misses": int((mask_true & (y_pred != code)).sum()),
        "false_positives": int((mask_pred & (y_true != code)).sum()),
        "precision": vals["precision"],
        "recall": vals["recall"],
        "f1": vals["f1-score"],
    })
metric_df = pd.DataFrame(metric_rows)

# Combine high-miss classes and low-recall classes so both large and fragile classes are represented.
miss_rank = metric_df.sort_values(["misses", "support"], ascending=[False, False]).head(12)["industry_code"].tolist()
recall_rank = metric_df[metric_df["support"] >= 20].sort_values(["recall", "support"], ascending=[True, False]).head(12)["industry_code"].tolist()
worst_codes = list(dict.fromkeys(miss_rank + recall_rank))[:18]

summary_rows = []
example_rows = []
for code in worst_codes:
    class_rows = raw_test[raw_test["true"] == code].copy()
    missed = class_rows[class_rows["pred"] != code].copy()
    predicted_as = raw_test[(raw_test["pred"] == code) & (raw_test["true"] != code)].copy()
    miss_pred_counts = missed["pred"].value_counts()
    fp_source_counts = predicted_as["true"].value_counts()
    miss_count = len(missed)
    same_sector = 0.0 if miss_count == 0 else np.mean([sector_map.get(p, "") == sector_map.get(code, "") for p in missed["pred"]])
    same_group = 0.0 if miss_count == 0 else np.mean([group_map.get(p, "") == group_map.get(code, "") for p in missed["pred"]])
    top_wrong = miss_pred_counts.index[0] if len(miss_pred_counts) else ""
    top_wrong_count = int(miss_pred_counts.iloc[0]) if len(miss_pred_counts) else 0
    top_wrong_share = top_wrong_count / miss_count if miss_count else 0.0
    top_fp_source = fp_source_counts.index[0] if len(fp_source_counts) else ""
    top_fp_count = int(fp_source_counts.iloc[0]) if len(fp_source_counts) else 0

    missing_desc_rate = float(class_rows["missing_segment_description"].mean()) if len(class_rows) else 0.0
    short_or_missing_rate = float(class_rows["missing_or_short_segment_description"].mean()) if len(class_rows) else 0.0
    largest_rate = float(class_rows["largest_share_bool"].mean()) if len(class_rows) else np.nan
    low_rev_rate = float((class_rows["revenue_share_num"] < 0.25).mean()) if class_rows["revenue_share_num"].notna().any() else np.nan
    avg_company_labels = float(class_rows["company_industry_label_count"].mean()) if len(class_rows) else 0.0

    tags = []
    notes = []
    industry_name = name_map.get(code, "")
    if len(class_rows) < 50:
        tags.append("low support")
        notes.append("few examples make the decision boundary unstable")
    if short_or_missing_rate >= 0.40:
        tags.append("missing/short segment text")
        notes.append("many rows lack enough segment-specific evidence")
    if same_group >= 0.45:
        tags.append("same-group boundary confusion")
        notes.append("most misses go to sibling labels with similar taxonomy definitions")
    elif same_sector >= 0.60:
        tags.append("same-sector ambiguity")
        notes.append("misses stay in the correct broad sector but choose the wrong industry")
    if largest_rate < 0.35:
        tags.append("non-largest/conglomerate context")
        notes.append("company profile can describe a broader company than the segment itself")
    if avg_company_labels >= 1.25:
        tags.append("multi-industry company rows")
        notes.append("same company appears with multiple industry labels in the test split")
    if top_wrong_share >= 0.25:
        tags.append("dominant confusion target")
        notes.append(f"largest wrong destination is {top_wrong} {name_map.get(top_wrong, '')}")
    if any(word in industry_name.lower() for word in ["conglomerate", "diversified", "specialty", "services"]):
        tags.append("broad/overlapping label name")
        notes.append("label wording overlaps with neighboring industries")
    if not tags:
        tags.append("mixed error pattern")
        notes.append("errors are spread across several destinations rather than one clear driver")

    row = metric_df.loc[metric_df["industry_code"] == code].iloc[0].to_dict()
    row.update({
        "miss_rate": 1 - row["recall"],
        "same_sector_miss_rate": same_sector,
        "same_group_miss_rate": same_group,
        "missing_segment_description_rate": missing_desc_rate,
        "missing_or_short_segment_description_rate": short_or_missing_rate,
        "median_segment_desc_chars": float(class_rows["seg_desc_len"].median()) if len(class_rows) else 0.0,
        "median_profile_chars": float(class_rows["profile_len"].median()) if len(class_rows) else 0.0,
        "largest_share_segment_rate": largest_rate,
        "low_revenue_share_rate": low_rev_rate,
        "avg_company_industry_labels": avg_company_labels,
        "top_wrong_prediction_code": top_wrong,
        "top_wrong_prediction_name": name_map.get(top_wrong, ""),
        "top_wrong_prediction_count": top_wrong_count,
        "top_wrong_prediction_share_of_misses": top_wrong_share,
        "top_false_positive_source_code": top_fp_source,
        "top_false_positive_source_name": name_map.get(top_fp_source, ""),
        "top_false_positive_source_count": top_fp_count,
        "cause_tags": " | ".join(dict.fromkeys(tags)),
        "cause_notes": " ".join(dict.fromkeys(notes)),
    })
    summary_rows.append(row)

    for _, r in missed.head(5).iterrows():
        example_rows.append({
            "true_code": code,
            "true_industry_name": industry_name,
            "pred_code": r["pred"],
            "pred_industry_name": name_map.get(r["pred"], ""),
            "CompanyId": r["CompanyId"],
            "AsOfDate": r["AsOfDate"],
            "SegmentName": r["SegmentName"],
            "SegmentDescription_excerpt": textwrap.shorten(r["SegmentDescription"], width=240, placeholder="..."),
            "LongProfile_excerpt": textwrap.shorten(r["LongProfile"], width=320, placeholder="..."),
            "revenue_share": r["revenue_share_num"],
            "is_largest_share_segment": r["largest_share_bool"],
        })

root_cause = pd.DataFrame(summary_rows).sort_values(["misses", "miss_rate"], ascending=[False, False])
miss_examples = pd.DataFrame(example_rows)
root_cause.to_csv(OUT / "task1_worst_class_root_cause_summary.csv", index=False)
miss_examples.to_csv(OUT / "task1_worst_class_missed_examples.csv", index=False)

display_cols = [
    "industry_code", "industry_name", "support", "misses", "recall", "f1",
    "same_group_miss_rate", "missing_or_short_segment_description_rate",
    "largest_share_segment_rate", "top_wrong_prediction_code", "top_wrong_prediction_name",
    "top_wrong_prediction_count", "cause_tags", "cause_notes",
]
print("Worst-class root-cause summary")
display(root_cause[display_cols].head(18))

plot_df = root_cause.head(15).sort_values("misses")
fig, ax = plt.subplots(figsize=(11, 7))
labels = plot_df["industry_code"] + " " + plot_df["industry_name"].fillna("")
ax.barh(labels, plot_df["misses"], color="#e45756")
ax.set_title("Task 1 worst classes: miss count")
ax.set_xlabel("Miss count")
ax.grid(axis="x", alpha=0.25)
fig.tight_layout()
fig.savefig(OUT / "task1_worst_class_miss_count.png", dpi=180)
plt.show()

plot_df = root_cause.head(15).sort_values("miss_rate")
fig, ax = plt.subplots(figsize=(11, 7))
labels = plot_df["industry_code"] + " " + plot_df["industry_name"].fillna("")
ax.barh(labels, plot_df["miss_rate"], color="#e45756", label="Miss rate")
ax.scatter(plot_df["same_group_miss_rate"], labels, color="#54a24b", label="Same-group miss rate", zorder=3)
ax.scatter(plot_df["missing_or_short_segment_description_rate"], labels, color="#f58518", label="Missing/short segment text rate", zorder=3)
ax.axvline(0.25, color="#777777", linestyle=":", linewidth=1.2, label="25% error reference")
ax.set_xlim(0, 1)
ax.set_title("Task 1 worst classes: likely error drivers")
ax.set_xlabel("Rate")
ax.grid(axis="x", alpha=0.25)
ax.legend(loc="lower right")
fig.tight_layout()
fig.savefig(OUT / "task1_worst_class_error_drivers.png", dpi=180)
plt.show()

print("Saved root-cause table:", OUT / "task1_worst_class_root_cause_summary.csv")
print("Saved missed-row examples:", OUT / "task1_worst_class_missed_examples.csv")
```

```python
# Top-10 industry performance for Task 1.
top10_industry_results = per_class[per_class["industry_code"].isin(top10)].copy()
top10_industry_results["rank_by_support"] = top10_industry_results["industry_code"].map({c: i + 1 for i, c in enumerate(top10)})
top10_industry_results["class_accuracy_recall"] = top10_industry_results["recall"]
top10_industry_results["correct"] = (top10_industry_results["class_accuracy_recall"] * top10_industry_results["support"]).round().astype(int)
top10_industry_results["misses"] = top10_industry_results["support"].astype(int) - top10_industry_results["correct"]
top10_industry_results = top10_industry_results.sort_values("rank_by_support")

top10_summary = {
    "top10_subset_accuracy": accuracy_score(y_true[top10_mask], y_pred[top10_mask]),
    "top10_macro_precision": float(top10_industry_results["precision"].mean()),
    "top10_macro_recall": float(top10_industry_results["recall"].mean()),
    "top10_macro_f1": f1_score(y_true[top10_mask], y_pred[top10_mask], labels=top10, average="macro", zero_division=0),
    "top10_weighted_f1": f1_score(y_true[top10_mask], y_pred[top10_mask], labels=top10, average="weighted", zero_division=0),
}

print("Task 1 top-10 industry summary")
display(pd.Series(top10_summary).to_frame("score"))
print("Per top-10 industry metrics")
display(top10_industry_results[[
    "rank_by_support", "industry_code", "industry_name", "support", "correct", "misses",
    "class_accuracy_recall", "precision", "recall", "f1"
]])

top10_industry_results.to_csv(OUT / "task1_top10_industry_metrics.csv", index=False)
pd.Series(top10_summary).to_csv(OUT / "task1_top10_industry_summary.csv")

fig, ax = plt.subplots(figsize=(10, 5.8))
plot_df = top10_industry_results.sort_values("rank_by_support")
ypos = np.arange(len(plot_df))
labels = plot_df["industry_code"] + " " + plot_df["industry_name"].fillna("")
ax.barh(ypos - 0.18, plot_df["f1"], height=0.36, label="F1", color="#4c78a8")
ax.barh(ypos + 0.18, plot_df["class_accuracy_recall"], height=0.36, label="Class accuracy / recall", color="#f58518")
ax.set_yticks(ypos)
ax.set_yticklabels(labels)
ax.invert_yaxis()
ax.set_xlim(0, 1)
ax.axvline(0.85, color="#d62728", linestyle="--", linewidth=1.4, label="0.85 top-10 target")
ax.set_title("Task 1: per-class performance for top-10 industries")
ax.set_xlabel("Score")
ax.grid(axis="x", alpha=0.25)
ax.legend()
fig.tight_layout()
fig.savefig(OUT / "task1_top10_industry_metrics.png", dpi=180)
plt.show()
```

```python
# Bottleneck view for Task 1 top-10 industries.
rows = []
for code in top10:
    mask_true = y_true == code
    mask_pred = y_pred == code
    misses_df = pd.DataFrame({"pred": y_pred[mask_true]})
    misses_df = misses_df[misses_df["pred"] != code]
    false_pos_df = pd.DataFrame({"true": y_true[mask_pred]})
    false_pos_df = false_pos_df[false_pos_df["true"] != code]

    top_wrong_pred = ""
    top_wrong_pred_name = ""
    top_wrong_pred_count = 0
    if len(misses_df):
        vc = misses_df["pred"].value_counts()
        top_wrong_pred = vc.index[0]
        top_wrong_pred_count = int(vc.iloc[0])
        top_wrong_pred_name = name_map.get(top_wrong_pred, "")

    top_false_source = ""
    top_false_source_name = ""
    top_false_source_count = 0
    if len(false_pos_df):
        vc = false_pos_df["true"].value_counts()
        top_false_source = vc.index[0]
        top_false_source_count = int(vc.iloc[0])
        top_false_source_name = name_map.get(top_false_source, "")

    rows.append({
        "industry_code": code,
        "industry_name": name_map.get(code, ""),
        "support": int(mask_true.sum()),
        "correct": int((y_pred[mask_true] == code).sum()),
        "misses": int((y_pred[mask_true] != code).sum()),
        "class_accuracy_recall": float((y_pred[mask_true] == code).mean()),
        "false_positives": int(((y_pred == code) & (y_true != code)).sum()),
        "top_wrong_prediction_code": top_wrong_pred,
        "top_wrong_prediction_name": top_wrong_pred_name,
        "top_wrong_prediction_count": top_wrong_pred_count,
        "top_false_positive_source_code": top_false_source,
        "top_false_positive_source_name": top_false_source_name,
        "top_false_positive_source_count": top_false_source_count,
    })

top10_bottlenecks = pd.DataFrame(rows).sort_values("misses", ascending=False)
top10_bottlenecks.to_csv(OUT / "task1_top10_industry_bottlenecks.csv", index=False)
display(top10_bottlenecks)

fig, ax = plt.subplots(figsize=(10, 5.8))
plot_df = top10_bottlenecks.sort_values("misses")
labels = plot_df["industry_code"] + " " + plot_df["industry_name"].fillna("")
ax.barh(labels, plot_df["misses"], color="#e45756")
ax.set_title("Task 1 bottleneck: misses among top-10 industries")
ax.set_xlabel("Miss count")
ax.grid(axis="x", alpha=0.25)
fig.tight_layout()
fig.savefig(OUT / "task1_top10_industry_bottlenecks.png", dpi=180)
plt.show()
```
